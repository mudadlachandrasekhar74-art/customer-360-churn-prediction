## Notebook 02: Customer 360 Feature Engineering & Behavioral Profiling

**Objective:** Aggregate the dimensional star schema to the unique natural person grain (`customer_unique_id`) to construct comprehensive RFM, behavioral, engagement, and operational customer features.

```python
import os
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DB_PATH = os.path.join('..', 'data', 'processed', 'olist_warehouse.db')
conn = sqlite3.connect(DB_PATH)
df_c360 = pd.read_sql('SELECT * FROM customer_360', conn)
conn.close()
print(f"Loaded customer_360: {df_c360.shape[0]:,} rows, {df_c360.shape[1]} features")
```

### 1. RFM Distributions (Recency, Frequency, Monetary Value)
Evaluating the skewness and concentration of customer engagement and monetary value.

```python
df_c360[['recency', 'frequency', 'monetary_value', 'average_order_value', 'average_review_score']].describe().round(2)
```

### 2. High-Value and Repeat Customer Concentrations

```python
repeat_rate = df_c360['repeat_customer_flag'].mean() * 100
hvc_rate = df_c360['high_value_customer_flag'].mean() * 100
print(f"Repeat Customer Rate: {repeat_rate:.2f}%")
print(f"High-Value Customer Rate (Top 20% Spend): {hvc_rate:.2f}%")
```
