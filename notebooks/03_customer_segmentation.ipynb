# Customer 360 Intelligence Platform
## Notebook 03: Customer Segmentation (RFM vs. K-Means Machine Learning)

**Objective:** Segment Olist's 96,096 customers using rule-based RFM matrices and unsupervised K-Means clustering to distinguish Champions, Loyalists, At-Risk high spenders, and experience-driven churn drivers.

```python
import os
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DB_PATH = os.path.join('..', 'data', 'processed', 'olist_warehouse.db')
conn = sqlite3.connect(DB_PATH)
df_c360 = pd.read_sql('SELECT * FROM customer_360', conn)
conn.close()
print(f"Loaded customer_360: {df_c360.shape[0]:,} customers with segmentation tags")
```

### 1. RFM Segment Distribution & Value Concentration

```python
rfm_summary = df_c360.groupby('rfm_segment').agg(
    Customer_Count=('customer_unique_id', 'count'),
    Total_Revenue=('monetary_value', 'sum'),
    Avg_Recency=('recency', 'mean'),
    Avg_Frequency=('frequency', 'mean'),
    Avg_AOV=('average_order_value', 'mean'),
    Avg_CSAT=('average_review_score', 'mean')
).sort_values('Total_Revenue', ascending=False).round(2)
rfm_summary['Revenue_Share_%'] = (rfm_summary['Total_Revenue'] / rfm_summary['Total_Revenue'].sum() * 100).round(2)
rfm_summary
```

### 2. K-Means Cluster Centroids & Profiling

```python
kmeans_summary = df_c360.groupby(['cluster_kmeans', 'cluster_label']).agg(
    Customer_Count=('customer_unique_id', 'count'),
    Total_Revenue=('monetary_value', 'sum'),
    Avg_Recency=('recency', 'mean'),
    Avg_Frequency=('frequency', 'mean'),
    Avg_AOV=('average_order_value', 'mean'),
    Avg_CSAT=('average_review_score', 'mean')
).round(2)
kmeans_summary
```
